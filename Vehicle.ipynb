```python
class Vehicle:
    def __init__(self, registration_number, model, fuel_type):
        self.registration_number = registration_number
        self.model = model
        self.fuel_type = fuel_type

    def show_details(self):
        print(f"Registration: {self.registration_number}")
        print(f"Model: {self.model}")
        print(f"Fuel Type: {self.fuel_type}")


class Car(Vehicle):
    def drive(self):
        print(f"{self.model} is driving on the road with passengers.")


class Bus(Vehicle):
    def carry_passengers(self, count):
        print(f"{self.model} is carrying {count} passengers.")


class Truck(Vehicle):
    def load_cargo(self, weight):
        print(f"{self.model} is loading {weight} kg of cargo.")


# ---- CAR ----
print("=== Enter Car Details ===")
car_reg   = input("Enter car registration number: ")
car_model = input("Enter car model: ")
car_fuel  = input("Enter car fuel type: ")

car = Car(car_reg, car_model, car_fuel)
print()
car.show_details()
car.drive()

print()

# ---- BUS ----
print("=== Enter Bus Details ===")
bus_reg   = input("Enter bus registration number: ")
bus_model = input("Enter bus model: ")
bus_fuel  = input("Enter bus fuel type: ")
bus_count = int(input("Enter number of passengers: "))

bus = Bus(bus_reg, bus_model, bus_fuel)
print()
bus.show_details()
bus.carry_passengers(bus_count)

print()

# ---- TRUCK ----
print("=== Enter Truck Details ===")
truck_reg    = input("Enter truck registration number: ")
truck_model  = input("Enter truck model: ")
truck_fuel   = input("Enter truck fuel type: ")
truck_weight = float(input("Enter cargo weight (kg): "))

truck = Truck(truck_reg, truck_model, truck_fuel)
print()
truck.show_details()
truck.load_cargo(truck_weight)
```
